```python
import pandas as pd
```

```python
df = pd.read_excel("HR. Project.xlsx", sheet_name="HR")
df
```

```python
# Drop useless columns
df = df.drop(columns=["EmployeeCount", "StandardHours", "Over18"])

# Drop id
df = df.drop(columns=[ "EmployeeNumber"])

# Clean text
df["BusinessTravel"] = df["BusinessTravel"].str.replace("_", " ")

# Remove duplicates
df = df.drop_duplicates()
```

```python
# save the file
df.to_excel("HR_Cleaned_1.xlsx", index=False)
```
